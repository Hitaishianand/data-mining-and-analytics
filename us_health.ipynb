Change Data Type of Float (that are actually categorical variables)


```python
cat=["education","currentSmoker","BP Meds","prevalentStroke","diabetes"]
```

```python
for i in cat:
  df[i]=df[i].astype("object")
```

```python
df.info()
```

```python
cat = []
num = []
for i in df.columns:
  if df[i].dtype == "object":
    cat.append(i)
  else:
    num.append(i)

print(cat)
print(num)
```

Rename columns with spaces in between their names

```python
df.rename(columns={"tot cholesterol":"tot_cholesterol","BP Meds": "BP_Meds","Systolic BP":"Systolic_BP","Diastolic BP":"Diastolic_BP"},inplace=True)
```

```python
df.info()
```

IMPUTATION OF MISSING VALUES

```python
df.Gender=df.Gender.fillna("Female")
```

```python
df.info()
```

```python
for column in ['education','BP_Meds','prevalentStroke','diabetes']:
  mode_val=df[column].mode()[0]
  df[column]=df[column].fillna(mode_val).astype("object")
```

```python
df.info()
```

```python
df.info()
```

```python
for column in ['age','cigsPerDay','tot_cholesterol','BMI','heartRate','glucose']:
  df[column]=df[column].fillna(df[column].median())
```
